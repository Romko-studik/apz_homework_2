### Analysis

Overall even though the completion time is longer in the latter tasks the overall implementations are still much better. We have to consider that for each request the number of services which are used is increased, which also increases the overall time. However, this massively increases the systems reselience and overall functionality - for the program from the first task even one service down is fatal. On the other hand we have safeguards for each service in the latest implementation, which is a great win considering only 15% slowdown.
