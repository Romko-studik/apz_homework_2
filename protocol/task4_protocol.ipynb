## 6. Fault Tolerance — counter-service paused

**POST while counter is paused — transactions still queue successfully:**

![pause post](./images/fault_pause_post.png)

**GET /accounts while counter is paused — returns null:**

![pause get](./images/fault_pause_get.png)

**After unpausing — counter drains queue, correct balances restored:**

![unpause](./images/fault_unpause.png)

### Analysis

- While counter-service is paused, POST requests still succeed — transactions accumulate in the Hazelcast Queue
- GET requests return null balance since counter-service is unreachable
- After resuming counter-service, it drains all queued transactions and balances become correct
- In conclusion - facade does not need to wait for counter-service, improving resilience and decoupling
