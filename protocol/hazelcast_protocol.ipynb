### 3.5 Data loss analysis

We see that if we stop 1 or 2 nodes sequentionaly there is no data loss. However if 2 nodes are stopped at the same time we lose some entries - we have 688 entries remaining out of 1000 initial. This can be prevented by increasing backup-count in hazelcast settings, as this way each entry will exist on more nodes. So if we increase backup-count to 2 each entry will have a backup the 2 remaining nodes. This way if we lose 2 out of 3 nodes simultaniously we still dont lose any data

## 7. Comparison

From the results of these 3 experiments we can clearly see that:
- Without locking we get race conditions resulting in loss of data - due to requests being skipped
- With both optimistic and pessimistic locking we get correct end value of 30000, however pessimistic locking works slightly faster
