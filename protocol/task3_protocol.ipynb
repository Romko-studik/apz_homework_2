### 4.5 Fault tolerance analysis

- 1 logging service down: facade falls back to remaining instances automatically
- 2 logging services down: facade falls back to the last remaining instance
- 1 Hazelcast node down: remaining nodes hold all data, system works
- 2 Hazelcast nodes down: all logging services lose connection (smart_routing=False means each instance only knows its assigned node), POST fails with 'All logging service instances unavailable', GET still works since MongoDB is unaffected

### Performance analysis

We can see that task 3 is slightly slower than task 1 but it's expected - since we are using MongoDB for user storage and Hazelcast for logging. Due to this the individual sppeed of each request execution is up to 250x slower, which is thankfully not that bad considering that it took a fraction of the whole time. Network avarage also slightly increased which is expected as well. However while this setup is working slower its much more robust - with better data handling and more redundancy in case a service fails, which was not there in the first task at all.
